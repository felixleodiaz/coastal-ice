```python
from osgeo import gdal
from pathlib import Path
from tqdm import tqdm
```

```python
# random sample of images

input_dir  = Path("../../local_data/RandomSampleImages")
output_dir = Path("../../local_data/RandomSampleImages_PNG")
output_dir.mkdir(exist_ok=True)

# do not create .aux.xml files for the PNGs

gdal.SetConfigOption('GDAL_PAM_ENABLED', 'NO')

# convert

for tif in tqdm(list(input_dir.glob("*.tif")), desc="Converting"):
    gdal.Translate(
        str(output_dir / f"{tif.stem}.png"),
        str(tif),
        format="PNG"
    )
```
